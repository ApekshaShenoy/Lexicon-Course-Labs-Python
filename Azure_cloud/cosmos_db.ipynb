```python
from azure.cosmos import CosmosClient

class CosmosService:

    def __init__(self, uri, key, database_name, container_name):
        client = CosmosClient(uri, credential=key)
        database = client.get_database_client(database_name)
        self.container = database.get_container_client(container_name)

    def create_item(self, data):
        return self.container.create_item(data)

    def get_items(self):
        return list(self.container.read_all_items())
```


**Why this data structure?**
Storing related product attributes together in a single document just makes sense here — it keeps things tidy and avoids the need for joins. It also plays naturally to the strengths of a document database, where the whole point is to model data the way you'd actually think about it.

**Why partition key `/category`?**
Category felt like a natural choice. It spreads products fairly evenly across partitions and lines up well with how people actually query the data — most of the time, you're filtering by category anyway, so it's a win on both ends.

**Challenges encountered**
Honestly, partitions took a little while to fully click. On top of that, getting the Python SDK connected and figuring out the right way to structure the JSON documents had its own learning curve — but working through those bumps made the whole thing much clearer.

**Scaling scenario**
One of the nicer things about Cosmos DB is that you don't have to think too hard about scaling. As demand grows, it automatically distributes data across partitions and adjusts throughput — it largely takes care of itself.

**Advantages over relational databases**
The flexibility is probably what stands out most. There's no rigid schema to wrestle with, and you get horizontal scalability and strong performance almost out of the box — especially for document-style data where a relational model would feel like a bit of a squeeze.
