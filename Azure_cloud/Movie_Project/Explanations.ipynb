# Cosmos DB Lab Part 2 — Django Integration

### Role of Cosmos DB

Cosmos DB is used to store movie data and reviews.

This data is suitable for NoSQL because:
- It contains nested structures (reviews inside movies)
- The schema is flexible and may evolve
- It avoids the need for joins

### Architecture

Data Flow:
User → Django → CosmosService → Cosmos DB → Response

Cosmos DB is used when:
- Fetching movies
- Storing new movies or reviews

### Django Integration

A service layer was created to interact with Cosmos DB.

The application retrieves movie data and exposes it through an API endpoint.

### Challenges

If Cosmos DB is unavailable:
- The application cannot retrieve data
- Error handling should be implemented

Data validation:
- Ratings must be between 1–5
- Required fields must be present

Consistency:
- Eventual consistency is acceptable for reviews

### Reflection

What worked well:
- Flexible data model
- Easy retrieval of nested data

What was difficult:
- Thinking in NoSQL
- Choosing partition key

When not to use Cosmos DB:
- When strong relationships are required

Scaling:
- Data is distributed across partitions

Possible improvements:
- Optimize partition key
- Duplicate data for faster reads
