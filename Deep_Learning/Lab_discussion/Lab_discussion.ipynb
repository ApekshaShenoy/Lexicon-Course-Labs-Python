# Lab Discussion: Object Detection and Generative AI

## 1. Why is object detection harder than image classification?

Image classification only answers **what is in the image**. The model predicts a single label for the entire image, such as “cat” or “car”.

Object detection is more complex because it must answer **two questions at the same time**:

1. **What objects are present**
2. **Where those objects are located**

The model must detect **multiple objects** and draw **bounding boxes** around each object while also classifying them. For example, in a street image the model may need to detect several cars, pedestrians, bicycles, and traffic lights all at different locations.

Because the model must both **localize and classify objects**, the task is much more computationally difficult than simple classification.



## 2. Why was the sliding window method inefficient for object detection?

The sliding window method worked by moving a **small window across the entire image** and running an image classifier on every position.

Problems with this method:

* It requires **thousands of evaluations** per image.
* The window must run at **multiple sizes and scales** to detect large and small objects.
* This results in **very high computational cost**.
* It is **too slow for real-time applications** like video processing or autonomous driving.

Example:
A single image might require **tens of thousands of window checks**, which makes the method extremely inefficient.


## 3. Why was YOLO considered a breakthrough for real-time systems?

YOLO (You Only Look Once) improved object detection by treating the problem as **a single neural network prediction** instead of scanning the image many times.

Key advantages:

* The image is processed **only once by the neural network**.
* The network **predicts bounding boxes and object classes simultaneously**.
* This makes the detection **much faster** than sliding window approaches.

Because of this design, YOLO can process images **in real time**, making it useful for applications such as:

* self-driving cars
* surveillance systems
* robotics
* video analytics



## 4. Why do generative models need huge datasets?

Generative models learn to **create new data that looks similar to the training data**.

To do this well, they must learn:

* patterns
* structures
* textures
* variations in the data

If the dataset is large, the model can learn **many examples and variations**, which helps it generate realistic results.

Examples:

* Image generation models need millions of images.
* Text models need billions of sentences.

Large datasets help the model **generalize and produce diverse outputs**.


## 5. What happens if the dataset is too small in generative models?

If the dataset is too small, several problems occur:

**Overfitting**
The model memorizes the training data instead of learning general patterns.

**Low diversity**
Generated outputs become repetitive and similar to the training data.

**Poor quality outputs**
Images or text may look unrealistic, distorted, or incomplete.

Example:
A generative model trained on only 100 face images may produce faces that look **very similar to the original training images** instead of new ones.

# Risks of Generative AI

## 1. Deepfakes

Generative AI can create realistic fake videos, images, or audio that appear to show real people saying or doing things they never did.

Risks include:

* impersonation
* political manipulation
* fraud
* damage to reputations

Example: Fake videos of public figures spreading false information.

## 2. Misinformation

Generative AI can produce large amounts of **false or misleading content quickly**.

This can spread through:

* social media
* fake news articles
* automated bots

The result can be confusion, manipulation of public opinion, or loss of trust in reliable information.


## 3. Copyright Issues

Generative models are trained on large datasets that may include **copyrighted materials** such as:

* artwork
* books
* music
* photographs

Problems arise when the AI generates content that is **very similar to the original creators’ work**, raising questions about:

* ownership
* intellectual property
* fair compensation for creators.
