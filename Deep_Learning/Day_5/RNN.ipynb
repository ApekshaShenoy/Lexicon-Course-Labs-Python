## R-CNN Pipeline Lab

### Introduction

Object detection is more challenging than simple image classification. In image classification, the model predicts a single label for the entire image. In object detection, the system must not only identify what objects are present but also determine where they are located in the image. Since an image can contain multiple objects of different sizes and positions, the task becomes more complex.

### Step 1: Region Proposals (Selective Search)

The first step in R-CNN is generating possible object locations using a method called Selective Search. Instead of checking every possible location in the image, this method groups similar pixels and generates around 2,000 region proposals that are likely to contain objects. This helps reduce the number of regions that need to be analyzed.

### Step 2: Cropping and Resizing Regions

Each proposed region is then cropped from the original image and resized to a fixed size. This step is necessary because the neural network requires all input images to have the same dimensions before processing.

### Step 3: Feature Extraction using CNN

The resized regions are passed through a Convolutional Neural Network such as AlexNet. The CNN extracts important visual features from the image regions, such as edges, shapes, and textures. These features help the model understand the visual patterns within each region.

### Step 4: Classification using SVM

After feature extraction, the features are given to Support Vector Machine classifiers. These classifiers determine whether the region contains a particular object or just background.

### Step 5: Bounding Box Regression

Once the object is identified, a bounding box regressor adjusts the predicted box so that it fits the object more accurately. This improves the precision of the object’s location in the image.

### Step 6: Non-Maximum Suppression

Since many region proposals overlap, the model may detect the same object multiple times. Non-Maximum Suppression removes these duplicate detections by keeping the bounding box with the highest confidence score and discarding the others.

### Conclusion

The R-CNN pipeline combines region proposals, deep learning feature extraction, and traditional machine learning classifiers to detect objects in an image. Although it can produce accurate results, it is computationally slow because each region proposal must be processed separately.

### Pipeline Sketch

Input Image
     ↓

     ↓

     ↓

     ↓

     ↓

     ↓

     ↓
