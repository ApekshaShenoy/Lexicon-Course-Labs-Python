### Task 1 — Zero Shot

In the first task, the AI was given almost no direction. The prompt was simply something like “build a simple Python chatbot.” Because of that, the result technically worked, but it felt like a rough first draft rather than something intentionally designed.

The code likely came out as one big block without much structure. There were probably no functions, very few or no comments, and no clear explanation of what certain parts were doing. It also didn’t feel personal. The chatbot didn’t know the user’s name, and the keyword responses were likely generic. If a beginner tried to read the code, they might struggle to understand things like `.lower()` or why `while True` was being used.

So while the program ran, it wasn’t really built with a specific user or learning purpose in mind.

**What was missing:** structure, comments, personalisation, and a clear target audience.

---

### Task 2 — Persona + Context

In the second task, the prompt added more guidance by telling the AI to act like a senior Python developer writing code for beginner students. That small change made a noticeable difference in the output.

Now the AI had a clearer sense of who the code was meant for. Because of that, the responses became friendlier, the chatbot greeted the user by name, and the keywords were more relevant to a student audience. The overall tone felt more approachable, and even the goodbye message sounded warmer instead of abrupt.

That said, the prompt still didn’t tell the AI how to structure the code. So it made its own decisions about formatting. It might have used functions, or it might not have. Some lines might have had comments while others didn’t. The final result was better, but still somewhat inconsistent.

**What improved:** the purpose, tone, personalisation, and relevance to the intended audience.

---

### Task 3 — Format

The third task is where the output really became useful as learning material. This time, the prompt included very specific formatting instructions: separate functions, a `main()` entry point, comments above every function, inline comments for confusing lines, and even a limit of 50 lines.

By doing this, the structure of the code was no longer left up to the AI’s guesswork. The prompt clearly defined what the final file should look like. As a result, the chatbot code became much easier to follow. A beginner could read it from top to bottom and understand what each part was doing.

The 50-line limit also helped keep things concise. It prevented unnecessary filler and forced the solution to stay focused and clear.

**What improved:** structure, readability, maintainability, and its usefulness as a teaching example.
