## Part 1 — Test the AI (Bias & Trust)

We observed a slight but noticeable bias when our prompts included names or details that carry social stereotypes. When we used a female name, the AI responded with a more uncertain tone and more cautious language compared to gender-neutral or male-named prompts — even though the qualifications were identical. In some cases the AI showed uncertainty, but it was not factually incorrect. Overall, we consider the AI's answers **partially trustworthy**: useful as a starting point, but not reliable enough to act on without human review.

## Part 2 — Build an AI Policy

**Scenario: AI News Generator**

**What it should be allowed to do**
- Summarise news from verified and credible sources
- Clearly label all content as AI-generated

**What it should NOT be allowed to do**
- Create or spread fake news or unverified claims
- Present opinions as facts
- Generate misleading or sensationalised headlines

**Handling bias**
- Draw from diverse and reputable sources
- Clearly separate factual reporting from analysis or opinion
- Conduct regular audits to identify and correct bias in output

**Handling user data**
- Collect only the minimum data necessary
- Keep all data secure and allow users to opt out or request deletion at any time
- Never sell user data to third parties

**If the AI makes a mistake**
- Correct and update the article quickly and transparently
- Allow users to report misinformation directly
- Maintain an AI-generated watermark on all content so readers always know the source

**Who is responsible**
Responsibility is shared between the platform hosting the AI, the developers who built it, and the human editors overseeing its output. No single party can pass accountability entirely to another.

## Part 3 — Prompt Engineering & Influence

Our key finding is that AI is highly prompt-sensitive — it is designed to follow the prompt closely and adapts its tone, framing, and conclusions based on the direction you set. Even small changes in wording can lead to meaningfully different output. This matters because it means AI can be deliberately steered toward producing biased, misleading, or one-sided content simply by how a question is phrased. The main risks this creates are manipulation and the spread of misinformation — especially when users are unaware that the AI is reflecting the prompt's assumptions rather than objective reality.

I hadn't expected something as small as a name to influence the output, and it made me realise that bias in AI isn't always obvious or dramatic. It can be subtle enough that most people would simply not notice it.
After this exercise, I trust AI more cautiously than before. I don't think AI is useless — it clearly has a lot of practical value — but I now understand that it can reflect and reinforce biases in ways that feel authoritative and neutral on the surface. That gap between how confident AI sounds and how reliable it actually is feels important to keep in mind.
Going forward, I will make a habit of cross-checking AI-generated answers with other sources before acting on them. If an AI gives me information that will influence a real decision, I won't take it at face value. This lab showed me that verifying AI output isn't just good practice — it's necessary.
